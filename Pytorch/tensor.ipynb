## Type conversion

```python
t1 = torch.tensor([1,2,3,4,5])
```

```python
t1 = t1.type(torch.float32)
print(t1.dtype)
```

```python
t1 = t1.to(torch.long)
print(t1.dtype)
```

```python
t1 = t1.float()
print(t1.dtype)
```

```python
# shared memory between torch tensor and numpy array
t1 = torch.tensor([1,2,3,4,5])
n1 = t1.numpy()
print(n1,type(n1))

t1[0] = 100 
print(t1,n1)  # n1 is also changed

# use copy to avoid shared memory
t1 = torch.tensor([1,2,3,4,5])
n1 = t1.numpy().copy()
print(n1,type(n1))
```

```python
n1 = np.array([1,2,3,4,5])
t1 = torch.from_numpy(n1).clone() # no shared memory
```

```python
t1 = torch.tensor([1,2,3,4,5])
print(t1[0],t1[0].item())  # convert to python scalar, item is only for single element tensor
```

```python
t2 = torch.tensor([[6]])
print(t2.item())  # convert to python scalar
```

## Basic Operation

```python
t1 = torch.tensor([10,20,30,40])
```

```python
t1 +10
```

```python
print(t1.mul(10))  # multiply by 10
print(t1)
print(t1.mul_(10))  # multiply by 10 in place
print(t1)  # t1 is changed
```

```python
t1 = torch.tensor([[10,20,30], [40,50,60]])
t2 = torch.tensor([[70,80,90], [100,110,120]])
```

```python
# element-wise addition
print(t1+t2)
```

```python
print(t1 * t2)  # element-wise multiplication
print(t1.mul(t2))  # element-wise multiplication
print(t1.mul_(t2))  # element-wise multiplication in place
print(t1)  # t1 is changed
```

```python
t1 = torch.tensor([[10,20,30], [40,50,60]])
t2 = torch.tensor([[70,80,90], [100,110,120]])
```

```python
print(t1 @ t2.T)
print(t1.mm(t2.T))  # matrix multiplication
print(t1.matmul(t2.T))  # matrix multiplication
```

```python
t1 = torch.tensor([[[10,20,30],[40,50,60]],[[10,20,30],[40,50,60]]])
print(t1,t1.shape)      
t2 = torch.tensor([[[70,80,90],[100,110,120],[130,140,150]],[[70,80,90],[100,110,120],[130,140,150]]])
print(t2,t2.shape)
```

```python
print(t1 @ t2)
print(t1.matmul(t2))  # matrix multiplication
```

## Statistical operation

```python
t1 = torch.tensor([[[10,20,30],[40,50,60]],[[10,20,30],[40,50,60]]])
print(t1,t1.shape)
```

```python
print(t1.sum())
```

```python
print(t1.sum(dim=0))  # sum along the first dimension
print(t1.sum(dim=1))  # sum along the second dimension
print(t1.sum(dim=2))  # sum along the third dimension
```

```python
t1 = t1.float()
print(t1.mean())
print(t1.mean(dim=0))  # mean along the first dimension
print(t1.mean(dim=1))  # mean along the second dimension    
```

```python
print(t1)
```

```python

print(t1.max())
print(t1.max(dim=0))  # max along the first dimension
```

```python
print(t1.unique())
```

```python
print(t1.sort())  # unique along the first dimension
```

## Indexing

```python
t1 = torch.randint(1,10,(2,5,4))
print(t1)
```

```python
print(t1[0,0,0])
print(t1[0][0][0])
```

```python
print(t1[1,2])
```

```python
print(t1[1,1:,2:])
```

```python
list1 = [0,1,1]
list2 = [1,3,2]
print(t1[list1,list2])  # advanced indexing
```

```python
mask = t1 >5
print(mask)
```

```python
print(t1[mask])  # boolean indexing
print(t1[t1>5])
```

```python
print(t1)
```

```python
mask = t1[:,:,0]>5
print(mask)
```

```python
print(t1[mask])  # boolean indexing
```

```python
mask = t1[:,0,:]>5
print(mask)
```

```python
print(t1.mT[mask].mT)  # boolean indexing
```

```python
t2 = t1.transpose(1,2)
print(t2)
print(t2 - t1.mT)
```

```python
t2 = t1.permute(0,2,1)
print(t2)
print(t2.shape)
```

```python
t2 = t1.reshape(2,2,10)
print(t1)
print(t2)
print(t2.shape)
```

```python

t2 = t1.view(2,-1)
print(t2)

print(t1.mT.is_contiguous())
# contiguous memory is required for view, but not for reshape
t2 = t1.mT.contiguous().view(2,-1)
# print(t2)
```

## Combine tensors

```python
t1 = torch.randint(1,10,(2,1,3))
t2 = torch.randint(1,10,(2,3,3))
print(t1)
print(t2)
```

```python
t3 = torch.cat((t1,t2),dim=1)
```

```python
t1 = torch.randint(1,10,(2,3,3))
t2 = torch.randint(1,10,(2,3,3))
t3 = torch.stack((t1,t2),dim=1)
```

```python
print(t3)
print(t3.shape)
```
