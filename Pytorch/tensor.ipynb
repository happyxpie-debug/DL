```python
import torch
import numpy as np
```

## Tensor creation
#### 1.1 Based on content

```python
t1 = torch.tensor(10)
print(t1,t1.shape)       
```

```python
t1 = torch.tensor([10,20,30])
print(t1,t1.shape)
```

```python
t1 = torch.tensor([[10,20,30],[40,50,60]])
print(t1,t1.shape)
```

```python
t1 = torch.tensor([[[10,20,30],[40,50,60]],[[10,20,30],[40,50,60]]])
print(t1,t1.shape)
```

#### 1.2 Base on shape, use Tensor class

```python
t1 = torch.Tensor(2,3,5)
print(t1,"\n",t1.shape)  # default float32
```

```python
t1 = torch.Tensor(2,3).int()
print(t1,"\n",t1.shape)
```

```python
# use specific data types   
t1 = torch.IntTensor(2,3)
print(t1.dtype)  
t1 = torch.LongTensor(2,3)
print(t1.dtype) 
t1 = torch.ShortTensor(2,3)
print(t1.dtype) 
t1 = torch.ByteTensor(2,3)
print(t1.dtype) 
```

```python
t1 = torch.DoubleTensor(2,3)
print(t1.dtype)
t1 = torch.FloatTensor(2,3)
print(t1.dtype)
t1 = torch.HalfTensor(2,3)
print(t1.dtype)
```

#### 1.3 based on range

```python
# similar to numpy arange
t1 = torch.arange(10,20,2)
print(t1)
t1 = torch.arange(10)
print(t1)
```

```python
t1 = torch.linspace(10,20,3)
print(t1)
```

```python
t1 = torch.logspace(1,3,5,base=np.e) # 5 points between 10^1 and 10^3
print(t1)
print(np.pow(np.e,3))
```

#### 1.4 fill in tensor with specific values

```python
t1 = torch.ones(3)
print(t1,t1.shape)
t1 = torch.ones_like(t1)
print(t1,t1.shape)

t1 = torch.empty(3)
print(t1,t1.shape)
t1 = torch.empty_like(t1)
print(t1,t1.shape)

t1 = torch.eye(3)
print(t1,t1.shape)

t1 = torch.full((3,3),7)
print(t1,t1.shape,t1.dtype)
```

## Type conversion

```python
t1 = torch.tensor([1,2,3,4,5])
```

```python
t1 = t1.type(torch.float32)
print(t1.dtype)
```

```python
t1 = t1.to(torch.long)
print(t1.dtype)
```

```python
t1 = t1.float()
print(t1.dtype)
```

```python
# shared memory between torch tensor and numpy array
t1 = torch.tensor([1,2,3,4,5])
n1 = t1.numpy()
print(n1,type(n1))

t1[0] = 100 
print(t1,n1)  # n1 is also changed

# use copy to avoid shared memory
t1 = torch.tensor([1,2,3,4,5])
n1 = t1.numpy().copy()
print(n1,type(n1))
```

```python
n1 = np.array([1,2,3,4,5])
t1 = torch.from_numpy(n1).clone() # no shared memory
```

```python
t1 = torch.tensor([1,2,3,4,5])
print(t1[0],t1[0].item())  # convert to python scalar, item is only for single element tensor
```

```python
t2 = torch.tensor([[6]])
print(t2.item())  # convert to python scalar
```
