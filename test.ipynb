### Vectorization speed test

```python
import time
a = np.random.rand(1000000)
b = np.random.rand(1000000)
start = time.time()
c = np.dot(a, b)
end = time.time()
print(f"Time taken: {end - start}")

c = 0
start = time.time()
for i in range(1000000):
    c += a[i] * b[i]
end = time.time()
print(f"Time taken: {end - start}")
```

### Multiply a vector by a matrix in NumPy, use np.dot(v, M) (for \(v^{T}M\)) or np.dot(M, v) (for Mv)

```python
v = np.array([1, 2])                  # Shape: (2,)
M = np.array([[3, 4, 5], [6, 7, 8]])  # Shape: (2, 3)

result = np.dot(v, M)  # Or: v @ M
result2 = v @ M
print(result)
print(result2)
```

### broadcasting

```python
A = np.random.rand(3, 4)
print(A)
```

```python
c = A.sum(axis=0) 
print(c,c.shape)
```

```python
p = A / c
print(p,p.shape)
```

```python
p = A / c.reshape(1,4)
print(p,p.shape,c.reshape(1,4).shape)
```
